# **Agente Reativo Simples**

**função** AGENTE-REATIVO-SIMPLES(percepção) **retorna** ação
  

> **dados:**
*   regras - *conjunto de regras se-então*



> estado = INTERPRETAR-ENTRADA(percepcao)

> regra = regras.buscar(estado)

> ação = regra.acao

> **devolver** ação

---



```python
ACOES = {
    'pedido-recebido': 'coletar-produto',
    'pacote-pronto': 'seguir-para-entrega',
    'endereco-encontrado': 'entregar-pacote',
    'entrega-concluida': 'retornar-ao-centro',
    'bateria-baixa': 'ir-para-recarga',
    'obstaculo': 'desviar-obstaculo'
}
```

```python
class AgenteTabular:
  #Agente racional do tipo Tabular

  def __init__(self,acoes):
    self.acoes = acoes
    self.percepcoes = ''

  def atuar(self, percepcao, acao_basica=''):
  #Atua de acordo com a percepção, devolvendo uma ação
    if not percepcao:
      return acao_basica
    if len(self.percepcoes) != 0:
      self.percepcoes += ','
    self.percepcoes += percepcao
    if self.percepcoes in self.acoes.keys():
      return self.acoes[self.percepcoes]
    self.percepcoes = ''
    return acao_basica
```

```python
print("-- Agente Tabular: Robô de Entregas da Amazon --")
maquina = AgenteTabular(ACOES)
percepcao = input("Indicar Percepção: ")
while percepcao:
  acao = maquina.atuar(percepcao,'reiniciar')
  print(acao)
  percepcao = input("Indicar Percepção: ")
```
