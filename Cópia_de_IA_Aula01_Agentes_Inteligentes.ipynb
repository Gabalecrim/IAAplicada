# **Agente Tabular**

**função** AGENTE-TABULAR(percepção) **retorna** ação
  

> **dados:**
*   percepções - *sequencia inicialmente vazia*
*   ações - *tabela indexada por percepções, totalmente especificada*


> perpecpções.agregar(percepção)

> ação = ações.buscar(percepções)

> **devolver** ação

```python
#Exemplo de Agente Inteligente Tabular

ACOES = { 'Chamar':'Aguerdando elevador chegar no seu andar',
'Chamar,T':'Elevador irá até o andar térreo',
'Chamar,1': 'Elevador irá até o andar 1° Andar',
'Chamar,2': 'Elevador irá até o andar 2° Andar',
'Chamar,T,Chamar': 'Aguerdando elevador chegar no seu andar',
'Chamar,1,Chamar': 'Aguerdando elevador chegar no seu andar',
'Chamar,2,Chamar': 'Aguerdando elevador chegar no seu andar',
'Chamar,T,Chamar,T': 'Elevador irá até o andar térreo',
'Chamar,T,Chamar,1': 'Elevador irá até o andar 1° Andar',
'Chamar,T,Chamar,2': 'Elevador irá até o andar 2° Andar',
'Chamar,1,Chamar,T': 'Elevador irá até o andar térreo',
'Chamar,1,Chamar,1': 'Elevador irá até o andar 1° Andar',
'Chamar,1,Chamar,2': 'Elevador irá até o andar 2° Andar',
'Chamar,2,Chamar,T': 'Elevador irá até o andar térreo',
'Chamar,2,Chamar,1': 'Elevador irá até o andar 1° Andar',
'Chamar,2,Chamar,2': 'Elevador irá até o andar 2° Andar'}
```

```python
class AgenteTabular:
  #Agente racional do tipo Tabular

  def __init__(self,acoes):
    self.acoes = acoes
    self.percepcoes = ''

  def atuar(self, percepcao, acao_basica=''):
  #Atua de acordo com a percepção, devolvendo uma ação
    if not percepcao:
      return acao_basica
    if len(self.percepcoes) != 0:
      self.percepcoes += ','
    self.percepcoes += percepcao
    if self.percepcoes in self.acoes.keys():
      return self.acoes[self.percepcoes]
    self.percepcoes = ''
    return acao_basica
```

```python
print("-- Agente Tabular: Elevador --")
Elevador = AgenteTabular(ACOES)
percepcao = input("Indicar Percepção: ")
while percepcao:
  acao = Elevador.atuar(percepcao,'reiniciar')
  print(acao)
  percepcao = input("Indicar Percepção: ")
```
