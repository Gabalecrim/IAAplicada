# **Agente Baseado em Modelos**

Armazena informação da parte do mundo que visitou (***memória***).
Mantém um ***estado interno*** e um ***modelo do mundo*** segundo uma sequência de percepções.


**função** AGENTE-BASEADO-MODELOS(percepção) **retorna** ação
  

> **dados:**
*   estado - *Descrição atual do mundo*
*   modelo - *Dado estado e ação, devolve um novo estado*
*   regras - *Conjunto de regras (se-então)*
*   ação - *Última ação realizada (inicialmente nenhuma)*

> estado = ATUALIZAR-ESTADO(estado, ação, percepcao, modelo)

> regra = regras.buscar(estado)

> ação = regra.acao

> **devolver** ação

---


```python
#Exemplo de Agente Inteligente Reativo Basado em Modelo

#Estados: sem-moeda, com-moeda, a1-servida, a2-servida, a3-servida
#Ações: pedir-moeda, pedir-codigo, esperar.
#Percepções: moeda, a1, a2, a3, servida

REGRAS = {
    'sem-pacote': 'aguardar-pacote',
    'com-pacote': 'realizar-entrega',
    'destino1-entregue': 'retornar-base',
    'destino2-entregue': 'retornar-base',
    'destino3-entregue': 'retornar-base'
  }

MODELO = {
    ('sem-pacote', 'aguardar-pacote', 'pacote'): 'com-pacote',

    ('com-pacote', 'realizar-entrega', 'destino1'):
        'destino1-entregue',

    ('com-pacote', 'realizar-entrega', 'destino2'):
        'destino2-entregue',

    ('com-pacote', 'realizar-entrega', 'destino3'):
        'destino3-entregue',

    ('destino1-entregue', 'retornar-base', 'entregue'):
        'sem-pacote',

    ('destino2-entregue', 'retornar-base', 'entregue'):
        'sem-pacote',

    ('destino3-entregue', 'retornar-base', 'entregue'):
        'sem-pacote'
}
```

```python
class AgenteReativoBaseadoModelo:
  #Agente racional do tipo reativo baseado em modelo

  def __init__(self,modelo, regras, estado_inicial='', acao_inicial=''):
    self.modelo = modelo
    self.regras = regras
    self.estado_inicial = estado_inicial
    self.acao_inicial = acao_inicial
    self.acao = None
    self.estado= self.estado_inicial
    self.ult_acao = self.acao_inicial

  def atuar(self, percepcao):
  #Atua de acordo com a percepção, devolvendo uma ação
    if not percepcao:
      return self.acao_inicial

    chave = (self.estado,self.ult_acao,percepcao)
    if chave not in self.modelo.keys():
      self.acao = None
      self.estado= self.estado_inicial
      self.ult_acao = self.acao_inicial
      return self.acao_inicial
    self.estado = self.modelo[chave]

    if self.estado not in self.regras.keys():
      self.acao = None
      self.estado= self.estado_inicial
      self.ult_acao = self.acao_inicial
      return self.acao_inicial
    acao = self.regras[self.estado]
    self.ult_acao = acao
    return acao
```

```python
print("-- Agente Reativo Baseado em Modelos: Drone de Entregas --")
drone = AgenteReativoBaseadoModelo(
  MODELO,
  REGRAS,
  'sem-pacote',
  'aguardar-pacote'
)
percepcao = input("Indicar Percepção: ")
while percepcao:
  acao = drone.atuar(percepcao)
  print(acao)
  percepcao = input("Indicar Percepção: ")
```

**(a) Mudanças do mundo**

O ambiente pode mudar sozinho. Por exemplo, um pacote pode ficar disponível para o drone ou o destino pode estar pronto para receber a entrega. O drone percebe essas mudanças por meio das percepções.

**(b) Efeito das ações**

As ações do drone também mudam o seu estado. Por exemplo, quando ele está sem pacote e recebe um pacote, passa para o estado `com-pacote`. Depois, ao realizar a entrega, passa para `destino1-entregue`. Após a entrega, ele retorna para a base e fica novamente no estado `sem-pacote`.

Assim, o drone usa o estado anterior, a última ação e a nova percepção para decidir qual será seu próximo estado.
