# **Agente Reativo Simples**

**função** AGENTE-REATIVO-SIMPLES(percepção) **retorna** ação
  

> **dados:**
*   regras - *conjunto de regras se-então*



> estado = INTERPRETAR-ENTRADA(percepcao)

> regra = regras.buscar(estado)

> ação = regra.acao

> **devolver** ação

---



```python
#Exemplo de Agente Inteligente Reativo

REGRAS = { 'moeda':'pedir-codigo',
          'a1': 'servir-bebida1',
          'a2': 'servir-bebida2',
          'a3': 'servir-bebida3'}
```

```python
class AgenteReativoSimples:
  #Agente racional do tipo reativo simples

  def __init__(self,regras):
    self.regras = regras

  def atuar(self, percepcao, acao_basica=''):
  #Atua de acordo com a percepção, devolvendo uma ação
    if not percepcao:
      return acao_basica

    if percepcao in self.regras.keys():
      return self.regras[percepcao]

    return acao_basica
```

```python
print("-- Agente Reativo Simples: Máquina de Bebidas --")
maquina = AgenteReativoSimples(REGRAS)
percepcao = input("Indicar Percepção: ")
while percepcao:
  acao = maquina.atuar(percepcao,'reiniciar')
  print(acao)
  percepcao = input("Indicar Percepção: ")
```
