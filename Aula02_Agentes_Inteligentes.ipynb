# **Agente Baseado em Modelos**

Armazena informação da parte do mundo que visitou (***memória***).
Mantém um ***estado interno*** e um ***modelo do mundo*** segundo uma sequência de percepções.


**função** AGENTE-BASEADO-MODELOS(percepção) **retorna** ação
  

> **dados:**
*   estado - *Descrição atual do mundo*
*   modelo - *Dado estado e ação, devolve um novo estado*
*   regras - *Conjunto de regras (se-então)*
*   ação - *Última ação realizada (inicialmente nenhuma)*

> estado = ATUALIZAR-ESTADO(estado, ação, percepcao, modelo)

> regra = regras.buscar(estado)

> ação = regra.acao

> **devolver** ação

---


```python
#Exemplo de Agente Inteligente Reativo Basado em Modelo

#Estados: sem-moeda, com-moeda, a1-servida, a2-servida, a3-servida
#Ações: pedir-moeda, pedir-codigo, esperar.
#Percepções: moeda, a1, a2, a3, servida

REGRAS = {'sem-moeda': 'pedir-moeda',
          'com-moeda': 'pedir-codigo',
          'a1-servida': 'esperar',
          'a2-servida': 'esperar',
          'a3-servida': 'esperar'}

MODELO = {('sem-moeda', 'pedir-moeda', 'moeda'): 'com-moeda',
          ('com-moeda', 'pedir-codigo', 'a1'): 'a1-servida',
          ('com-moeda', 'pedir-codigo', 'a2'): 'a2-servida',
          ('com-moeda', 'pedir-codigo', 'a3'): 'a3-servida',
          ('a1-servida', 'esperar', 'servida'): 'sem-moeda',
          ('a2-servida', 'esperar', 'servida'): 'sem-moeda',
          ('a3-servida', 'esperar', 'servida'): 'sem-moeda'}
```

```python
class AgenteReativoBaseadoModelo:
  #Agente racional do tipo reativo baseado em modelo

  def __init__(self,modelo, regras, estado_inicial='', acao_inicial=''):
    self.modelo = modelo
    self.regras = regras
    self.estado_inicial = estado_inicial
    self.acao_inicial = acao_inicial
    self.acao = None
    self.estado= self.estado_inicial
    self.ult_acao = self.acao_inicial

  def atuar(self, percepcao):
  #Atua de acordo com a percepção, devolvendo uma ação
    if not percepcao:
      return self.acao_inicial

    chave = (self.estado,self.ult_acao,percepcao)
    if chave not in self.modelo.keys():
      self.acao = None
      self.estado= self.estado_inicial
      self.ult_acao = self.acao_inicial
      return self.acao_inicial
    self.estado = self.modelo[chave]

    if self.estado not in self.regras.keys():
      self.acao = None
      self.estado= self.estado_inicial
      self.ult_acao = self.acao_inicial
      return self.acao_inicial
    acao = self.regras[self.estado]
    self.ult_acao = acao
    return acao
```

```python
print("-- Agente Reativo Baseado em Modelos: Máquina de Bebidas --")
maquina = AgenteReativoBaseadoModelo(MODELO, REGRAS, 'sem-moeda','pedir-moeda')
percepcao = input("Indicar Percepção: ")
while percepcao:
  acao = maquina.atuar(percepcao)
  print(acao)
  percepcao = input("Indicar Percepção: ")
```
